```python
import numpy as np
import matplotlib.pyplot as plt
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.neighbors import KNeighborsClassifier
from sklearn.metrics import accuracy_score

# Generate a 2D dataset
X, y = make_classification(n_samples=200, n_features=2, n_informative=2,
                           n_redundant=0, n_clusters_per_class=1, random_state=42)

# Visualize the dataset
plt.figure(figsize=(8, 6))
plt.scatter(X[:, 0], X[:, 1], c=y, cmap='viridis', s=50, edgecolors='k')
plt.title('Synthetic 2D Dataset')
plt.xlabel('Feature 1')
plt.ylabel('Feature 2')
plt.grid(True)
plt.show()
```

```python
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)
```

```python
# Initialize KNN Classifier with Euclidean distance
knn_euclidean = KNeighborsClassifier(n_neighbors=5, metric='euclidean')

# Train the model
knn_euclidean.fit(X_train, y_train)

# Make predictions
y_pred_euclidean = knn_euclidean.predict(X_test)

# Calculate accuracy
accuracy_euclidean = accuracy_score(y_test, y_pred_euclidean)
print(f"KNN with Euclidean Distance Accuracy: {accuracy_euclidean:.4f}")

# Initialize KNN Classifier with Manhattan distance
knn_manhattan = KNeighborsClassifier(n_neighbors=5, metric='manhattan')

# Train the model
knn_manhattan.fit(X_train, y_train)

# Make predictions
y_pred_manhattan = knn_manhattan.predict(X_test)

# Calculate accuracy
accuracy_manhattan = accuracy_score(y_test, y_pred_manhattan)
print(f"KNN with Manhattan Distance Accuracy: {accuracy_manhattan:.4f}")

# Compare results
if accuracy_euclidean > accuracy_manhattan:
    print("\nEuclidean distance performed better.")
elif accuracy_manhattan > accuracy_euclidean:
    print("\nManhattan distance performed better.")
else:
    print("\nBoth distance metrics performed equally.")
```
